# Review auxiliary experiment — *EOlife X*

### Overview
This notebook provides an **interactive interface** for reviewing the auxiliary experiment data used in the publication on **the validation of *EOlife X***.  
The auxiliary experiment involved a single subject ventilated across three conditions:
regular ventilation, intra-arrest mechanical ventilation, and intra-arrest manual ventilation (bag-valve).

Three data sources are overlaid for each signal:

| Track | Source |
|-------|--------|
| **OCR step-function** | Screen-recording forward-fill (`vi`, `ve`, `f` channels) |
| **EOlife stored** | Per-breath CSV export (`Vi`, `Vt`, `f` labels; n=3 rolling average for `f`) |
| **Flow reference** | `VTinsp`, `VTexp`, `Respiratory Rate` derived from the airflow signal |
| **Per-breath f (CSV)** | Raw `Freq (/min)` column timestamped at end-of-breath (`f (CSV)` label) |

### Usage Instructions
- The **overview strip** at the bottom is clickable — click to jump to that time window.  
- Press **`+`** / **`-`** to zoom, **arrow keys** to pan.  
- Scroll the **mouse wheel** over a subplot to scale its y-axis.

```python
from vitabel import Vitals
from pathlib import Path
import pandas as pd

DATA_PATH = Path("./data/aux_X0001126.json")

case = Vitals()
case.load_data(DATA_PATH)

present = set(case.get_label_names())

vi_labels = [l for l in ["Vi", "VTinsp"]                         if l in present]
ve_labels = [l for l in ["Vt", "VTexp"]                          if l in present]
f_labels  = [l for l in ["f",  "Respiratory Rate", "f (CSV)"]    if l in present]

case.plot_interactive(
    channels=[
        ["vi", "Vi (displayed)"],
        ["ve", "Vt (displayed)", "V_exp"],
        ["f",  "f (displayed)"],
        ["Flow"],
        ["pAw"],
    ],
    labels=[
        vi_labels+["Inspiration", "Expiration"],
        ve_labels+["Inspiration", "Expiration"],
        f_labels+["Inspiration", "Expiration"],
        ["Inspiration", "Expiration"],
        ["Inspiration", "Expiration"],
    ],
    channel_overviews=[["vi"]],
    subplots_kwargs={"figsize": (18, 13)},
    time_unit="s",
)
```
